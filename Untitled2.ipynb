```python
from ConfigSpace import ConfigurationSpace
from ray import tune
from smac import HyperparameterOptimizationFacade, Scenario


class Smac3Searcher(tune.search.Searcher):
    def __init__(s, smac):
        super().__init__()
        s.smac = smac
        s.inf = {}

    def suggest(s, id):
        s.inf[id] = s.smac.ask()
        return s.inf[id].config

    def on_trial_complete(s, id, result):
        s.smac.tell(s.inf[id], -result["a"])
```

```python
def train(x, seed=0):
    return -x["a"] ** 2


tuner = tune.Tuner(
    train,
    tune_config=tune.TuneConfig(
        "max",
        "a",
        Smac3Searcher(HyperparameterOptimizationFacade(Scenario(ConfigurationSpace({"a": (-1, 1)})), train)),
        tune.schedulers.ASHAScheduler(),
        100,
        2,
    ),
)
tuner.fit();
```
