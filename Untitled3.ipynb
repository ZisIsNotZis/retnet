```python
import numpy as np
import torch
from torch.nn import *
from functools import reduce

class Lut:
    def __init__(s, a):
        s._a = a
        s._A = np.lib.stride_tricks.sliding_window_view(np.log(a), 2).mean(1)
        s._add = s(a + a[:,None])
        s._sub = s((a - a[:,None]).clip(1e-45))
        s._0 = s(1.)
    def __call__(s, x):
        x = np.asarray(x)
        return s._a[x] if x.dtype==int else np.searchsorted(s._A, np.log(x))
    def add(s, a, b):
        return s._add[a, b]
    def sub(s, a, b):
        return s._sub[a, b]
    def mul(s, a, b):
        return (a + b - s._0).clip(0, len(s._A))
    def div(s, a, b):
        return (a - b + s._0).clip(0, len(s._A))
    def sum(s, a):
        return reduce(s.add, a.transpose(a.ndim-1,*range(a.ndim-1)))
    def prod(s, a):
        return (a.sum(-1) - s._0 * (a.shape[-1]  -1)).clip(0, len(s._A))
    def matmul(s, a, b):
        return s.sum(s.mul(a[...,None,:], b[...,None,:,:]))

@type.__call__
class Lut:
    def __init__(s, x=6e-5, y=6e4, n=65536):
        s.x = np.log(x)
        s.y = np.log(y)
        s.n = n
        s.o = s(torch.tensor(1))
    def __call__(s, x):
        return (x.sign() * x.abs().log().sub(s.x).mul(s.n / (s.y - s.x))).clamp(1 - s.n, s.n - 1)#.round()
    def __getitem__(s, x):
        return x.sign() * x.abs().mul((s.y - s.x) / s.n).add(s.x).exp()
    def add(s, x, y):
        return s(s[x] + s[y])
    def sub(s, x, y):
        return s(s[x] - s[y])
    def mul(s, x, y):
        return (x.sign() * y.sign() * (x.abs() + y.abs() - s.o)).clamp(1 - s.n, s.n - 1)
    def div(s, x, y):
        return (x.sign() * y.sign() * (x.abs() - y.abs() + s.o)).clamp(1 - s.n, s.n - 1)
    def sum(s, x):
        return s(s[x].sum(-1))
    def prod(s, x):
        return (x.sum(-1) - s.o * (x.size(-1) - 1)).clamp(1 - s.n, s.n - 1)
    def dot(s, x, y):
        return s.sum(s.mul(x[...,None,:], y[...,None,:,:]))
```
