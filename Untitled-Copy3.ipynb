```python
from functools import reduce

import numpy as np
import torch
from torch.nn import *
from tqdm.notebook import tqdm


class F(Module):
    def __init__(s, f, *_, **__):
        super().__init__()
        s.f = f
        s._ = _
        s.__ = __
        for _ in enumerate(_), __.items():
            for _, __ in _:
                if isinstance(__, Parameter):
                    s.register_parameter(str(_), __)
                elif isinstance(__, Module):
                    s.register_module(str(_), __)

    def forward(s, x):
        return s.f(x, *s._, **s.__)

    def _get_name(s):
        s = getattr(s.f, "__name__", str(s.f))
        return s[0].upper() + s[1:]

    def extra_repr(s):
        return ", ".join([*[str(tuple(s.shape) if isinstance(s, Parameter) else s) for s in s._], *[f"{s}={_}" for s, _ in s.__.items()]])


class BatchNorm(BatchNorm1d):
    def forward(s, x):
        return super().forward(x.flatten(end_dim=-2)).unflatten(0, x.shape[:-1])


class Add(Sequential):
    f = torch.add

    def __new__(s, *_, **__):
        _ = [_ for _ in _ if _]
        if len(_) < 2:
            return _[0]
        return object.__new__(s)

    def __init__(s, *_):
        super().__init__()
        for i, _ in enumerate(_):
            super().register_module(f"{type(s).__name__[0]}{i}", _)

    def forward(s, x):
        x = [s(x) for s in s]
        try:
            return reduce(s.f, x)
        except:
            print(*[x.shape for x in x])
            raise


class Mul(Add):
    f = torch.mul


class Dot(Add):
    f = torch.matmul


def Seq(*_):
    _ = [_ for _ in _ if _ and not isinstance(_, Identity) for _ in (_ if type(_) == Sequential else [_])]
    return Sequential(*_) if len(_) > 1 else _[0] if _ else Identity()


def Conv1dPadLeft(i, o, r, stride=1, padding=0, dilation=1, groups=1, bias=False, neck=0):
    if neck >= i * r or neck >= o * r:
        neck = 0
    if neck:
        return Seq(
            Conv1d(i, neck, 1, groups=groups, bias=False),
            r > 1 and Conv1d(neck, neck, r, stride, padding, dilation, neck, False),
            padding and F(torch.Tensor.__getitem__, (..., slice(-padding))),
            Conv1d(neck, o, 1, groups=groups, bias=False),
        )
    return Seq(
        Conv1d(i, o, r, stride, padding, dilation, groups, False),
        padding and F(torch.Tensor.__getitem__, (..., slice(-padding))),
    )


def MaxPool1dPadLeft(r, stride=None, padding=0, dilation=1):
    return Seq(
        padding and F(functional.pad, (padding, 0)),
        MaxPool1d(r, stride, 0, dilation),
    )


def decay(x, *k, dim=-1, mod="avg", shift=1, logspace=False):
    if shift:
        x = functional.pad(x[(*[slice(None)] * (x.ndim + dim), slice(-shift))], [*[0, 0] * (-dim - 1), shift, 0])
    k = reduce(torch.add, k)
    d = torch.arange(x.size(dim) / 2, -x.size(dim) / 2, -1, dtype=x.dtype, device=x.device)[(..., *[None] * (-dim - 1))]
    if logspace:
        f = torch.logcumsumexp if mod != "max" else lambda *_: torch.cummax(*_)[0]
        d = k * d
        D = d.logcumsumexp(dim) if mod == "sum" else d
        return x.sign().mul_(f(x.abs_().add_(5e-41).log_().add_(d), dim).sub_(D).exp_())
    f = torch.Tensor.cumsum_ if mod != "max" else lambda *_: torch.cummax(*_)[0]
    d = k.exp_() ** d
    D = d.cumsum(dim) if mod == "sum" else d
    return f(x.mul_(d), dim).div_(D)


def score(acc, p):
    return np.log1p(-p) - (-acc).log1p()


class DatBin:
    def __init__(s, a, l, *bs):
        s.a = np.memmap(a)
        s.unmap = np.zeros(256, "?")
        s.unmap[s.a] = True
        s.unmap = s.unmap.nonzero()[0]
        s.map = np.empty(256, "l")
        s.map[s.unmap] = range(len(s.unmap))
        s.a = torch.from_numpy(s.a).as_strided((len(s.a) - l, l + 1), (1, 1))
        s.map = torch.from_numpy(s.map)
        s.bs = bs

    def __call__(s):
        return s.map[s.a[torch.randint(len(s.a), s.bs)].long()].to("cuda", torch.long)


T = F(torch.transpose, -2, -1)
torch.set_printoptions(linewidth=288)
dat = DatBin("py.txt", 32, 256)
```

```python
def decore(i, o, k, min=2, max=6, **_):
    return Seq(
        Conv1d(i, k, 1, bias=False),
        F(decay, Parameter(init.uniform_(torch.empty(k, 1), np.log(0.1) / min, np.log(0.1) / max)), **_),
        Conv1d(k, o, 1, bias=False),
    )


def retcore(i, o, k, neck=[0] * 5, r=[1, 1, 1, 0, 1], h=1, ful=False, ret=True, min=2, max=6, **_):
    return Seq(
        Dot(
            Seq(
                Conv1dPadLeft(i, h * k[0], r[0], padding=r[0] - 1, neck=h * neck[0]),
                T,
                Unflatten(-1, [h, 1, k[0]]),
            ),
            Seq(
                Seq(
                    Conv1dPadLeft(i, h * k[0] * k[1], r[1], padding=r[1] - 1, neck=h * neck[1]),
                    T,
                    Unflatten(-1, [h, *k]),
                )
                if ful
                else Mul(
                    Seq(
                        Conv1dPadLeft(i, h * k[0], r[1], padding=r[1] - 1, neck=h * neck[1]),
                        Unflatten(-2, [h, k[0], 1]),
                    ),
                    Seq(
                        Conv1dPadLeft(i, h * k[1], r[2], padding=r[2] - 1, bias=False, neck=h * neck[2]),
                        Unflatten(-2, [h, 1, k[1]]),
                    ),
                ),
                r[3]
                and Seq(
                    Flatten(-3),
                    Conv1dPadLeft(h * k[0] * k[1], h * k[0] * k[1], r[3], padding=r[3] - 1, groups=h, bias=False, neck=h * neck[3]),
                    Unflatten(-1, [h, *k]),
                ),
                ret
                and F(
                    decay,
                    Parameter(init.uniform_(torch.empty(h, k[0], 1, 1), np.log(0.3) / min, np.log(0.3) / max)),
                    Parameter(init.uniform_(torch.empty(h, 1, k[1], 1), np.log(0.3) / min, np.log(0.3) / max)),
                    **_,
                ),
                F(torch.permute, (0, 4, 1, 2, 3)),
            ),
        ),
        Flatten(-3),
        T,
        Conv1dPadLeft(h * k[1], o, r[4], padding=r[4] - 1, bias=False, neck=h * neck[4]),
    )


def retcorelin(i, o, k, v, neck=0, h=1, min=3, max=7):
    return Seq(
        Dot(
            Seq(
                Linear(i, h * k),
                Unflatten(-1, [h, 1, k]),
            ),
            Seq(
                Mul(
                    Seq(
                        Linear(i, h * k),
                        Unflatten(-1, [h, 1, k]),
                    ),
                    Seq(
                        Linear(i, h * v, False),
                        Unflatten(-1, [h, v, 1]),
                    ),
                ),
                F(decay, Parameter(init.uniform_(torch.empty(h, 1, k), np.log(0.3) / min, np.log(0.3) / max)), Parameter(init.uniform_(torch.empty(h, v, 1), np.log(0.3) / min, np.log(0.3) / max)), dim=-4),
                T,
            ),
        ),
        Flatten(-3),
        neck and Linear(h * v, neck, False),
        Linear(neck or h * v, o, False),
    )


def covcore(i, o, k, r=3, ful=False):
    return Seq(
        not ful or k < i and Conv1d(i, k, 1, bias=False),
        Conv1d(min(k, i) if ful else k, o if ful else k, r, padding=r - 1, groups=1 if ful else k, bias=False),
        F(torch.Tensor.__getitem__, (..., slice(None, 1 - r))),
        not ful and Conv1d(k, o, 1, bias=False),
    )


def polcore(i, o, k, r=2, f=AvgPool1d):
    Seq(
        Conv1d(i, k, 1, bias=False),
        F(functional.pad, (r - 1, 0)),
        f(r, 1),
        Conv1d(k, o, 1, bias=False),
    )


def model(m, e=1, h=1, r=1, n=2):
    return Seq(
        Embedding(len(dat.unmap), m),
        T,
        *[
            Add(
                Seq(
                    Add(
                        retcore(m, m * e, [m * e] * 2, neck=[0, 0, 0, 0, m], h=h),
                        Conv1dPadLeft(m, m * e, r, padding=r - 1, bias=False),
                    ),
                    BatchNorm1d(m * e, track_running_stats=False),
                    SiLU(True),
                    Conv1d(m * e, m if n else len(dat.unmap), 1, bias=False),
                ),
                Conv1d(m, m if n else len(dat.unmap), 1, bias=not n),
            )
            for n in range(1 - n, 1)
        ],
    ).to("cuda", torch.bfloat16)


def modelin(m, e=1, h=1, r=1, n=2):
    return Seq(
        Embedding(len(dat.unmap), m),
        *[
            Add(
                Seq(
                    Add(
                        retcorelin(m, m * e, m * e, m * e, m, h),
                        Linear(m, m * e),
                    ),
                    T,
                    BatchNorm1d(m * e, track_running_stats=False),
                    T,
                    SiLU(True),
                    Linear(m * e, m if n else len(dat.unmap), False),
                ),
                Linear(m, m if n else len(dat.unmap), not n),
            )
            for n in range(1 - n, 1)
        ],
    ).to("cuda", torch.bfloat16)
```

```python
m = [modelin(64, h=8, n=1)]
sz = torch.tensor([sum(m.numel() for m in m.parameters()) for m in m])
print(sz, m[-1])
o = torch.optim.Adam((m for m in m for m in m.parameters()), eps=5e-41)
O = torch.optim.lr_scheduler.OneCycleLR(o, 0.01, 10000, pct_start=0.01, div_factor=4, final_div_factor=9)
X = dat()
L = A = 0
for i in tqdm(range(1, 1 + O.total_steps)):
    x, X = X, dat()
    y = [m(x[:, :-1]) for m in m]
    x = x[:, 1:]
    l = torch.stack([functional.cross_entropy(y.mT, x) for y in y])
    l.mean().backward()
    o.step()
    o.zero_grad()
    O.step()
    L += l.detach().float()
    A += torch.stack([(y.argmax(2) == x).sum(0) for y in y])
    if not i % 100:
        print("%.0f" % L, (A / len(x)).to("cpu", torch.long))
        L = A = 0
```
