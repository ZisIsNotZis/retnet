def isort():
    p = yield
    if p is None:
        return
    c = 1
    l = r = None
    i = yield
    while i is not None:
        if i < p:
            if l is None:
                l = isort()
                l.send(None)
            i = yield l.send(i)
            continue
        if i > p:
            if r is None:
                r = isort()
                r.send(None)
            i = yield r.send(i)
            continue
        c += 1
        i = yield
    if l is not None:
        yield from l
    yield p, c
    if r is not None:
        yield from r

```python
from math import log


def searchsorted(a, v):
    l = 0
    r = len(a) - 1
    while r >= l:
        i = (l + r) // 2
        t = a[i]
        if v < t:
            r = i - 1
        elif v > t:
            l = i + 1
        else:
            return i, None
    return l, True


def isort():
    c = {}
    for i in range(1024):
        i = yield
        if i is None:
            break
        c[i] = c.get(i, 0) + 1
    p = sorted(c)
    c = [c[i] for i in p]
    s = [None] * (len(p) + 1)
    if i is not None:
        i = yield
    while i is not None:
        j, _ = searchsorted(p, i)
        if _ is None:
            c[j] += 1
        else:
            _ = s[j]
            if _ is None:
                _ = s[j] = isort()
                _.send(None)
            _.send(i)
        i = yield
    for i, j, t in zip(s, p, c):
        if i is not None:
            yield from i
        yield j, t
    if s[-1] is not None:
        yield from s[-1]


def entropy(a, n):
    c = 0
    for _, a in a:
        c += a * log(a)
    return log(n) - c / n
```

```python
s = isort()
s.send(None)
for i in range(100):
    s.send(i)
entropy(s, 100)
```
