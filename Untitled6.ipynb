```python
from asyncio import Future, Task


class Futures:
    f = None

    def __init__(s, *a, n=0, b=9e9):
        s.a = []
        s.n = len(a) + n
        s.b = b
        for a in a:
            Task(a).add_done_callback(s.set_result)

    def set_result(s, a):
        try:
            a = getattr(a, "result", lambda: a)()
        except Exception as e:
            a = e
        if a == StopAsyncIteration:
            s.n = 0
            if s.f and not s.f.done():
                s.f.set_result(s.a)
                s.a = []
        else:
            s.a += [a]
            s.n = max(s.n - 1, 0)
            if len(s.a) == s.b or not s.n and s.f and not s.f.done():
                s.f.set_result(s.a)
                s.a = []

    def __await__(s):
        f = Future()
        if len(s.a) < s.b and s.n:
            s.f = f
        elif s.a:
            f.set_result(s.a)
            s.a = []
        else:
            f.set_exception(StopAsyncIteration)
        return f.__await__()

    def __aiter__(s):
        return s

    def __anext__(s):
        return s
```
