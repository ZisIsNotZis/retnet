```python
from inspect import signature

import numpy as np

f = dict(
    asort=lambda a, **_: a.argsort().view("I")[::2],
    sort=lambda a, asort=None, **_: np.sort(a) if asort is None else a[asort],
    sort_nemsk=lambda sort, **_: sort[:-1] != sort[1:],
    sort_ne=lambda sort_nemsk, **_: np.r_[sort_nemsk.nonzero()[0].view("I")[::2], len(sort_nemsk)],
    uniq=lambda sort, sort_ne, **_: sort[sort_ne],
    uniq_id=lambda asort, sort_ne, **_: asort[sort_ne],
    uniq_ct=lambda sort_ne, **_: np.r_[sort_ne[0] + 1, sort_ne[1:] - sort_ne[:-1]],
    uniq_inv=lambda asort, uniq_ct, **_: asort[np.arange(len(uniq_ct), dtype="I").repeat(uniq_ct)],
    uniq_n=lambda sort_nemsk, **_: sort_nemsk.sum() + 1,
)


def calc(f, *wanted, **given):
    want = list(wanted)
    calc = set(wanted)
    for i in want:
        for i, t in signature(f[i]).parameters.items():
            if i not in given and i != "_":
                want.append(i)
                if t.default:
                    calc.add(i)
    for i in want[::-1]:
        if i not in given and i in calc:
            given[i] = f[i](**given)
    t = [given[i] for i in wanted]
    return t if len(t) > 1 else t[0]


def predictability(a):
    inv, ct = calc(f, "uniq", "uniq_ct", a=a)
    inv = calc(f, "uniq_inv", sort=inv["f0"], asort=type("", (), {"__getitem__": lambda s, x: x})())
    t = np.zeros(inv[-1] + 1, "I")
    np.maximum.at(t, inv, ct)
    return t.sum() / len(a)
```
