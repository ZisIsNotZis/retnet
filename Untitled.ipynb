```python
import numpy as np
import torch
from torch.nn import *


def decay(x, cache={}):
    d, D = cache.get(x.shape[-3:], (None, None))
    if d is None:
        d = (torch.logspace(-1e-3, -1e-2, x.size(-1)) * torch.logspace(-1e-3, -1e-2, x.size(-2))[:, None]) ** torch.arange(x.size(-3), 0.0, -1)[:, None, None]
        D = d.cumsum(0)
        cache[x.shape[-3:]] = d, D
    return x.mul_(d).cumsum_(-3).div_(D)


class F(Module):
    def __init__(s, f, *_):
        super().__init__()
        s.f = f
        s._ = _

    def forward(s, x):
        return s.f(x, *s._)


class BatchNorm(BatchNorm1d):
    def forward(s, x):
        return super().forward(x.flatten(end_dim=-2)).unflatten(0, x.shape[:-1])


class Add(Sequential):
    def forward(s, x):
        return s[0](x).add_(s[1](x))


class Mul(Sequential):
    def forward(s, x):
        return s[0](x) * s[1](x)


class Dot(Sequential):
    def forward(s, x):
        return s[0](x) @ s[1](x).mT


class Dat:
    unmap = b" abcdefghijklmnopqrstuvwxyz0123456789_.\"&|^+\\-*/%@=()[]{}<>:\n\t~#,'"
    map = torch.empty(127, dtype=torch.long)
    map[list(unmap)] = torch.arange(len(unmap))
    a = torch.from_numpy(np.memmap("py.txt"))
    a = a.as_strided((len(a) - 256, 257), (1, 1))

    @classmethod
    def f(s, bs=56):
        return s.map[s.a[torch.randint(len(s.a), [56])].long()]
```

```python
m = 256
m = Sequential(
    Embedding(len(Dat.unmap), m),
    *[
        Sequential(
            Add(
                Dot(
                    Linear(m, m),
                    Sequential(
                        Mul(
                            Linear(m, m),
                            Sequential(
                                Linear(m, m, False),
                                F(torch.transpose, -2, -1),
                            ),
                        ),
                        F(decay),
                    ),
                ),
                Linear(m, m, False),
            ),
            Add(
                Sequential(
                    Linear(m, m, False),
                    BatchNorm(m, track_running_stats=False),
                    SiLU(True),
                ),
                Linear(m, m, False),
            ),
        )
        for _ in range(2)
    ],
    Linear(m, len(Dat.unmap)),
)
o = torch.optim.Adam(m.parameters())
O = torch.optim.lr_scheduler.OneCycleLR(o, 0.0005, 20000, pct_start=0.1, div_factor=9, final_div_factor=99)
L = A = 0
for i in range(O.total_steps):
    x = Dat.f(60)
    y = m(x[:, :-1, None])[:, :, 0]
    x = x[:, 1:]
    l = functional.cross_entropy(y.mT, x)
    l.backward()
    o.step()
    o.zero_grad()
    O.step()
    print(end=f"{l*100:.0f}_{(y.argmax(2)==x).sum()*100/x.numel():.0f} ")
```
