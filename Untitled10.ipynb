```python
from signal import default_int_handler, signal

import matplotlib.pyplot as plt
from IPython.display import clear_output
from nn import *


class Conv(Conv1d):
    def forward(s, x):
        return super().forward(x.mT).mT[:, : x.size(1) // s.stride[0]]


class ConvTranspose(ConvTranspose1d):
    def forward(s, x):
        return functional.pad(super().forward(x.mT).mT[:, : x.size(1) * s.stride[0] - 1], (0, 0, 1, 0))


class BatchNorm(BatchNorm1d):
    def forward(s, x):
        return super().forward(x.mT).mT


def Log(*A, N=100):
    X = n = 0
    a = []
    A = np.empty((0,) + A, "f")

    def Log(x=None):
        nonlocal X, n, A
        if x is None:
            if a:
                A = np.concatenate((A, np.stack(a) / N))
                a.clear()
            return A
        X += x
        n += 1
        if n >= N:
            a.append(X.to("cpu", torch.float, True).numpy())
            X = n = 0

    return Log


dat = Dat("../dat/py.txt", 1, 1024, 65)
```

```python
def Lin(i, o=0, b=False):
    return Linear(i, o or i, b)


def Ba(i):
    return Seq(BatchNorm(i, track_running_stats=False), SiLU(True))


def Cov(n, i, o=0):
    return Seq(*(Seq(Conv(i, o or i, 3, padding=2, bias=False), Ba(o or i)) for n in range(n)))


def Covd(n, i, o=0):
    return Seq(Conv(i, o or i, 3, 2, 1, bias=False), Ba(o or i), Cov(n, o or i))


def Covu(n, i, o=0):
    return Seq(Cov(n, i), ConvTranspose(i, o or i, 3, 2, bias=False), Ba(o or i))


def Ret(i, h=0, k=0, v=0, o=0):
    h = h or (i / 4).__ceil__()
    k = k or (i / h).__ceil__()
    v = v or k
    return Seq(
        Dot(
            Seq(Lin(i, h * k, True), Unflatten(-1, (h, 1, k))),
            Seq(
                Mul(
                    Seq(Lin(i, h * k, True), Unflatten(-1, (h, 1, k))),
                    Seq(Lin(i, h * v, True), Unflatten(-1, (h, v, 1))),
                ),
                Decay((h, 1, k), (h, v, 1), d=-4, shift=1),
            ),
        ),
        Flatten(-3),
        Lin(h * v, o or i),
        Ba(o or i),
    )


m, n = 64, 3
m = Seq(
    Add(Seq(Embedding(68, m, scale_grad_by_freq=True), Cov(n, m)), Embedding(68, m, scale_grad_by_freq=True)),
    Ret(m),
    Add(Cov(n, m), Lin(m)),
    Add(
        Seq(
            Add(Covd(n, m), Covd(0, m)[0]),
            Ret(m),
            Add(Cov(n, m), Lin(m)),
            Add(
                Seq(
                    Add(Covd(n, m), Covd(0, m)[0]),
                    Ret(m),
                    Add(Covu(n, m), Covu(0, m)[0]),
                ),
                Lin(m),
            ),
            Add(Cov(n, m), Lin(m)),
            Ret(m),
            Add(Covu(n, m), Covu(0, m)[0]),
        ),
        Lin(m),
    ),
    Add(Cov(n, m), Lin(m)),
    Ret(m),
    Add(Seq(Cov(n, m), Lin(m, 68)), Lin(m, 68, True)),
)
print(m)
m = torch.compile(torch.jit.trace(m.to("cuda", non_blocking=True), dat()[:, :-1]))
o = torch.optim.Adam(m.parameters(), 1e-3)
s = torch.cuda.amp.GradScaler()
L = 9e9
logl = Log()
loga = Log()
sigint = False
signal(2, lambda *_: globals().__setitem__("sigint", True))
for i, x in enumerate(dat):
    o.zero_grad()
    with torch.autocast("cuda", torch.bfloat16):
        y = m(x[:, :-1])
        x = x[:, 1:]
        l = functional.cross_entropy(y.mT, x)
    s.scale(l).backward()
    s.step(o)
    s.update()
    logl(l.detach())
    loga((y.argmax(2) != x).sum() / x.numel())
    if not (i + 1) % 1000 or sigint:
        plt.subplot(211).plot(logl(), label=f"{logl().min()*1000:.0f} {logl()[-1]*1000:.0f}")
        plt.yscale("log")
        plt.legend()
        plt.subplot(212).plot(loga(), label=f"{loga().min()*1000:.0f} {loga()[-1]*1000:.0f}")
        plt.yscale("log")
        plt.legend()
        clear_output(True)
        plt.show()
        if sigint:
            break
signal(2, default_int_handler)
```
