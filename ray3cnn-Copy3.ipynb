```python
from time import time

import matplotlib.pyplot as plt
from IPython.display import clear_output
from ipywidgets import Output
from nn import *


def f(k, d, h, n=2, b=256, l=32, lr=0.001, div=(0.01, 9, 99), t=60, type=torch.float32, out=None):
    a = Dat("../dat/py.txt", 0, b, l)
    m = Seq(
        Embedding(68, k),
        *(
            Res(
                Mul(
                    Lin(k, d, True),
                    Seq(
                        Mul(
                            Lin(k, d, True),
                            Lin(k, d, True),
                        ),
                        Decay(d, max=32),
                    ),
                ),
                Lin(d, h, True),
                ReLU(True),
                Lin(h, k),
            )
            for _ in range(n)
        ),
        Lin(k, 68, True),
    ).to("cuda", type, True)
    m = torch.compile(torch.jit.trace(m, a()[:, :-1]))
    m(a()[:, :-1])
    o = torch.optim.Adam(m.parameters())
    O = torch.optim.lr_scheduler.OneCycleLR(o, lr, t * 10, pct_start=div[0], div_factor=div[1], final_div_factor=div[2])
    L = Log(100)
    A = Log(100)
    t = time()
    T = 0
    while time() - t < O.total_steps / 10:
        x = a()
        y = m(x[:, :-1])
        x = x[:, 1:]
        l = functional.cross_entropy(y.mT, x)
        o.zero_grad()
        l.backward()
        o.step()
        L(l)
        A((y.argmax(2) != x).sum() / x.numel())
        if time() - t > O.last_epoch / 10 + 0.1:
            O.step()
            if out and time() - t > T * 20:
                plt.plot(L())
                plt.twinx().plot(A(), c="orange")
                plt.grid()
                out.clear_output(True)
                with out:
                    plt.show()
                T += 1
    print(k, d, h, L()[-1], A()[-1])
    return L()[-1]


def find(**a):
    I = next(iter(a))
    out = Output()
    display(out)
    while a[I] <= 2048:
        L = 9e9
        for i in a:
            if (l := f(**a | {i: a[i] * 2}, out=out)) < L:
                L, I = l, i
        a[I] *= 2
    return a


def Log(N=1, f=lambda x: x, F=lambda x: x):
    X = n = 0
    a = []
    A = torch.empty(0)

    def Log(x=None):
        nonlocal X, n, A
        if x is None:
            if a:
                A = torch.cat((A, F(torch.stack(a).mul_(1 / N)).to("cpu", non_blocking=True)))
                a.clear()
            return A
        X += f(x)
        n += 1
        if n >= N:
            a.append(X)
            X = n = 0

    return Log
```
