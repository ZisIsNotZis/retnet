    d = cache.get(key)
    if d is None:
        A = torch.linspace((A - a) / 2, (B - b) / 2, x.size(-2))
        a = torch.linspace(a / 2, b / 2, x.size(-2))
        d = (a + a[:, None] + (A + A[:, None]) * torch.linspace(0, 1, x.size(-3))[:, None, None]).to(x)
        d = cache[key] = d, d.logcumsumexp(0)
    return x.sign().mul_(x.abs_().add_(5e-41).log_().add_(d[0]).logcumsumexp(-3).sub(d[1]).exp_())

```python
from functools import reduce

import numpy as np
import torch
from torch.nn import *
from tqdm.notebook import tqdm


def cumsumdecay(x, k, logspace=True):
    d = torch.arange(x.size(-1) / 2, -x.size(-1) / 2, -1, dtype=x.dtype, device=x.device)
    if logspace:
        d = k * d
        return x.sign().mul_(x.abs_().add_(5e-41).log_().add_(d).logcumsumexp(-1).sub(d.logcumsumexp(-1)).exp_())
    d = k.exp() ** d
    return x.mul_(d).cumsum_(-1).div_(d.cumsum(-1))


def cummaxdecay(x, k, logspace=True):
    d = torch.arange(x.size(-1) / 2, -x.size(-1) / 2, -1, dtype=x.dtype, device=x.device)
    if logspace:
        d = k * d
        return x.sign().mul_(x.abs_().add_(5e-41).log_().add_(d).cummax(-1)[0].sub(d).exp_())
    d = k.exp() ** d
    return x.mul_(d).cummax(-1)[0].div_(d)


class F(Module):
    def __init__(s, f, *_, **__):
        super().__init__()
        s.f = f
        s._ = _
        s.__ = __
        for _ in enumerate(_), __.items():
            for _, __ in _:
                if isinstance(__, Parameter):
                    s.register_parameter(str(_), __)
                elif isinstance(__, Module):
                    s.register_module(str(_), __)

    def forward(s, x):
        return s.f(x, *s._, **s.__)

    def extra_repr(s):
        return ", ".join(str(s.f), *(map(str, s._), *(f"{i}={j}" for i, j in s.__.items())))


class BatchNorm(BatchNorm1d):
    def forward(s, x):
        return super().forward(x.flatten(end_dim=-2)).unflatten(0, x.shape[:-1])


def Seq(*_):
    return Sequential(*[_ for _ in _ if _ and not isinstance(_, Identity) for _ in (_ if type(_) == Sequential else [_])])


class Add(Sequential):
    f = torch.add

    def __new__(s, *_):
        _ = [_ for _ in _ if _]
        if len(_) < 2:
            return _[0]
        return object.__new__(s)

    def __init__(s, *_):
        super().__init__(*[_ for _ in _ if _])

    def forward(s, x):
        return reduce(s.f, [s(x) for s in s])


class Mul(Add):
    f = torch.mul


class Dot(Add):
    f = lambda a, b: a @ b.T


def MaxPool1dPadLeft(r, s=None, p=0, d=1):
    return Seq(
        p and F(functional.pad, (p, 0)),
        MaxPool1d(r, s, 0, d),
    )


class Dat:
    def __init__(s, a, l, *bs):
        s.a = np.array([open(a).read()]).view("I")
        s.unmap = np.zeros(65536, "?")
        s.unmap[s.a] = True
        s.unmap = s.unmap.nonzero()[0].astype("I")
        s.map = np.empty(65536, "h")
        s.map[s.unmap] = range(len(s.unmap))
        s.a = torch.from_numpy(s.map[s.a]).as_strided((len(s.a) - l, l + 1), (1, 1))
        s.bs = bs

    def __call__(s):
        return s.a[torch.randint(len(s.a), s.bs)].to("cuda", torch.long)


class DatBin:
    def __init__(s, a, l, *bs):
        s.a = np.memmap(a)
        s.unmap = np.zeros(256, "?")
        s.unmap[s.a] = True
        s.unmap = s.unmap.nonzero()[0]
        s.map = torch.empty(256, dtype=torch.uint8)
        s.map[s.unmap] = torch.arange(len(s.unmap), dtype=torch.uint8)
        s.a = torch.from_numpy(s.a).as_strided((len(s.a) - l, l + 1), (1, 1))
        s.bs = bs

    def __call__(s):
        return s.map[s.a[torch.randint(len(s.a), s.bs)].long()].to("cuda", torch.long)


def Model(e, k, K, d, decay=lambda k: F(decay)):
    return Seq(
        F(torch.unsqueeze, -1),
        Embedding(e, K),
        *[
            Add(
                Seq(
                    Dot(
                        Linear(K, k),
                        Seq(
                            Mul(
                                Linear(K, k),
                                Seq(
                                    Linear(K, k, False),
                                    F(torch.transpose, -2, -1),
                                ),
                            ),
                            decay(k),
                        ),
                    ),
                    Linear(k, K, False),
                    BatchNorm(K),
                    SiLU(True),
                ),
                Linear(K, K, False),
            )
            for i in range(d)
        ],
        Linear(K, e),
        F(lambda x: x.squeeze(-2).mT),
    ).to("cuda", torch.bfloat16)


def ModelC(k, r=3, R=2, enc=[1], dec=[1], nn=2, fullin=True, prepoolin=True, res=True, poolres=True, add=Add, bias=False, postpoolin=False, dilate=True, g=1, d=1):
    return Seq(
        *[
            add(
                Seq(
                    *[
                        Seq(
                            Conv1d(k, k, r, 1, (r - 1) * d, d, k, False),
                            F(torch.Tensor.__getitem__, (..., slice(None, (1 - r) * d))),
                            BatchNorm1d(k),
                            SiLU(True),
                            Conv1d(k, k, 1, groups=g, bias=False),
                            BatchNorm1d(k),
                            SiLU(True),
                        )
                        for i in range(i)
                    ],
                    fullin and Conv1d(k, k, 1, groups=g, bias=bias),
                ),
                res and Conv1d(k, k, 1, groups=g, bias=bias),
            )
            for i in enc
        ],
        nn
        and add(
            Seq(
                prepoolin and Conv1d(k, k, 1, groups=g, bias=False),
                MaxPool1dPadLeft(R, 1 if dilate else R, (R - 1) * d, d),
                ModelC(k, r, R, enc, dec, nn - 1, fullin, prepoolin, res, poolres, add, bias, postpoolin, g, d * R),
                postpoolin and Conv1d(k, k, 1, groups=g, bias=bias),
                not dilate and F(lambda x: functional.pad(x.repeat_interleave(R, -1)[..., :-1], (1, 0))),
            ),
            poolres and Conv1d(k, k, 1, groups=g, bias=bias),
        ),
        *[
            add(
                Seq(
                    *[
                        Seq(
                            Conv1d(k, k, r, 1, (r - 1) * d, d, k, False),
                            F(torch.Tensor.__getitem__, (..., slice(None, (1 - r) * d))),
                            BatchNorm1d(k),
                            SiLU(True),
                            Conv1d(k, k, 1, groups=g, bias=False),
                            BatchNorm1d(k),
                            SiLU(True),
                        )
                        for i in range(i)
                    ],
                    fullin and Conv1d(k, k, 1, groups=g, bias=bias),
                ),
                res and Conv1d(k, k, 1, groups=g, bias=bias),
            )
            for i in (dec if nn else [])
        ],
    )


dat = DatBin("cat", 4096, 16)
```

```python
def ModelC(k, enc=[[[3]], [[3]], [[3]]], pool=[2, 2], dec=[[[3]], [[3]], []], fullin=False, prepoolin=False, add=Add, bias=False, postpoolin=False, dilate=True, g=1, d=1):
    return Seq(
        *[
            add(
                Seq(
                    *[
                        Seq(
                            Conv1d(k, k, i, 1, (i - 1) * d, d, k, False),
                            F(torch.Tensor.__getitem__, (..., slice(None, (1 - i) * d))),
                            BatchNorm1d(k),
                            SiLU(True),
                            Conv1d(k, k, 1, groups=g, bias=False),
                            BatchNorm1d(k),
                            SiLU(True),
                        )
                        for i in i
                    ],
                    fullin and Conv1d(k, k, 1, groups=g, bias=bias),
                ),
                Conv1d(k, k, 1, groups=g, bias=bias),
            )
            for i in enc[0]
        ],
        pool
        and add(
            Seq(
                prepoolin and Conv1d(k, k, 1, groups=g, bias=False),
                MaxPool1dPadLeft(pool[0], 1 if dilate else pool[0], (pool[0] - 1) * d, d),
                ModelC(**{**locals(), "enc": enc[1:], "pool": pool[1:], "dec": dec[1:], "d": d * pool[0]}),
                postpoolin and Conv1d(k, k, 1, groups=g, bias=bias),
                not dilate and F(lambda x: functional.pad(x.repeat_interleave(pool[0], -1)[..., :-1], (1, 0))),
            ),
            Conv1d(k, k, 1, groups=g, bias=bias),
        ),
        *[
            add(
                Seq(
                    *[
                        Seq(
                            Conv1d(k, k, i, 1, (i - 1) * d, d, k, False),
                            F(torch.Tensor.__getitem__, (..., slice(None, (1 - i) * d))),
                            BatchNorm1d(k),
                            SiLU(True),
                            Conv1d(k, k, 1, groups=g, bias=False),
                            BatchNorm1d(k),
                            SiLU(True),
                        )
                        for i in i
                    ],
                    fullin and Conv1d(k, k, 1, groups=g, bias=bias),
                ),
                Conv1d(k, k, 1, groups=g, bias=bias),
            )
            for i in dec[0]
        ],
    )
```
