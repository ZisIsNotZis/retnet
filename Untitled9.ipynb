```python
from signal import default_int_handler, signal

import matplotlib.pyplot as plt
from IPython.display import clear_output
from nn import *


class Conv(Conv1d):
    def forward(s, x):
        return super().forward(x.mT).mT[:, : x.size(1) // s.stride[0]]


class ConvTranspose(ConvTranspose1d):
    def forward(s, x):
        return functional.pad(super().forward(x.mT).mT[:, : x.size(1) * s.stride[0] - 1], (0, 0, 1, 0))


class BatchNorm(BatchNorm1d):
    def forward(s, x):
        return super().forward(x.mT).mT


def Log(*A, N=100):
    X = n = 0
    a = []
    A = np.empty((0,) + A, "f")

    def Log(x=None):
        nonlocal X, n, A
        if x is None:
            if a:
                A = np.concatenate((A, np.stack(a) / N))
                a.clear()
            return A
        X += x
        n += 1
        if n >= N:
            a.append(X.to("cpu", torch.float, True).numpy())
            X = n = 0

    return Log


def Lin(i, o=0, b=False):
    return Linear(i, o or i, b)


def Ba(i):
    return Seq(
        BatchNorm(i, track_running_stats=False),
        SiLU(True),
    )


def Cov(i, n):
    return Seq(
        *[
            Seq(
                Conv(i, i, 3, padding=2, bias=False),
                Ba(i),
            )
            for n in range(n)
        ]
    )


def Covs(i, *n, before=lambda: 0, after=lambda b: 0):
    return Seq(
        *[
            Add(
                Seq(
                    before(),
                    Cov(i, n),
                    after(False),
                ),
                before() or after(True) or Lin(i),
            )
            for n in n
        ]
    )


def Covd(i, *n):
    return Seq(
        Covs(i, *n[:-1]),
        Add(
            Seq(
                Cov(i, n[-1]),
                Conv(i, i, 3, 2, 1, bias=False),
                Ba(i),
            ),
            Conv(i, i, 3, 2, 1, bias=False),
        ),
    )


def Covu(i, *n):
    return Seq(
        Add(
            Seq(
                ConvTranspose(i, i, 3, 2, bias=False),
                Ba(i),
                Cov(i, n[0]),
            ),
            ConvTranspose(i, i, 3, 2, bias=False),
        ),
        Covs(i, *n[1:]),
    )


dat = Dat("../dat/py.txt", 0.01, 512, 129)


def Ret(i, h=0, k=0, v=0):
    h = h or (i / 4).__ceil__()
    k = k or (i / h).__ceil__()
    v = v or k
    return Seq(
        Dot(
            Seq(Lin(i, h * k, True), Unflatten(-1, (h, 1, k))),
            Seq(
                Mul(
                    Seq(Lin(i, h * k, True), Unflatten(-1, (h, 1, k))),
                    Seq(Lin(i, h * v, True), Unflatten(-1, (h, v, 1))),
                ),
                Decay((h, 1, k), (h, v, 1), d=-4, shift=1, min=8, max=32),
            ),
        ),
        Flatten(-3),
        Lin(h * v, i),
        Ba(i),
    )
```

```python
def Ret(i, h=0, k=0, v=0, **_):
    h = h or (i / 4).__ceil__()
    k = k or (i / h).__ceil__()
    v = v or k
    return Add(
        Seq(
            Add(
                Seq(
                    Dot(
                        Seq(Lin(i, h * k, True), Unflatten(-1, (h, 1, k))),
                        Seq(
                            Mul(
                                Seq(Lin(i, h * k, True), Unflatten(-1, (h, 1, k))),
                                Seq(Lin(i, h * v, True), Unflatten(-1, (h, v, 1))),
                            ),
                            Decay((h, 1, k), (h, v, 1), d=-4, shift=1, **_),
                        ),
                    ),
                    Flatten(-3),
                    Lin(h * v, i),
                ),
                Lin(i, i),
            ),
            Ba(i),
        ),
        Lin(i, i),
    )


m, r, c = 64, [16], [3]
m = Seq(
    Covs(m, *c, before=lambda: Embedding(68, m, scale_grad_by_freq=True)),
    *[
        Seq(
            Ret(m, *r),
            Covs(m, *c),
            Add(
                Seq(
                    Covd(m, *c),
                    Ret(m, *r),
                    Covs(m, *c),
                    Add(
                        Seq(
                            Covd(m, *c),
                            Ret(m, *r),
                            Covs(m, *c),
                            Add(
                                Seq(
                                    Covd(m, *c),
                                    Ret(m, *r),
                                    Covs(m, *c),
                                    Add(
                                        Seq(
                                            Covd(m, *c),
                                            Ret(m, *r),
                                            Covu(m, *c),
                                        ),
                                        Lin(m),
                                    ),
                                    Covs(m, *c),
                                    Ret(m, *r),
                                    Covu(m, *c),
                                ),
                                Lin(m),
                            ),
                            Covs(m, *c),
                            Ret(m, *r),
                            Covu(m, *c),
                        ),
                        Lin(m),
                    ),
                    Covs(m, *c),
                    Ret(m, *r),
                    Covu(m, *c),
                ),
                Lin(m),
            ),
            Covs(m, *c),
        )
        for _ in range(1)
    ],
    Ret(m, *r),
    Covs(m, *c, after=lambda b: Lin(m, 68, b)),
)
print(m)
m = torch.compile(torch.jit.trace(m.to("cuda", non_blocking=True), dat()[:, :-1]))
o = torch.optim.Adam(m.parameters(), 1e-3)
s = torch.cuda.amp.GradScaler()
L = 9e9
logl = Log()
loga = Log()
sigint = False
signal(2, lambda *_: globals().__setitem__("sigint", True))
for i, x in enumerate(dat):
    o.zero_grad()
    with torch.autocast("cuda", torch.bfloat16):
        y = m(x[..., :-1])
        x = x[..., 1:]
        l = functional.cross_entropy(y.mT, x)
    s.scale(l).backward()
    s.step(o)
    s.update()
    logl(l.data)
    loga((y.argmax(-1) != x).sum() / x.numel())
    if not (i + 1) % 100 or sigint:
        plt.subplot(211).plot(logl())
        plt.yscale("log")
        plt.legend([(logl()[-1] * 1000).astype("H").astype("U4")])
        plt.subplot(212).plot(loga())
        plt.yscale("log")
        plt.legend([(loga()[-1] * 1000).astype("H").astype("U3")])
        clear_output(True)
        plt.show()
        if sigint:
            break
signal(2, default_int_handler)
```

```python
from bisect import bisect
from signal import ITIMER_REAL, SIGALRM, setitimer, signal
from time import time

DEFER = []
DEFERF = None


def defer(x, f):
    global DEFERF
    if DEFERF:
        x += time()
        DEFER.insert(bisect(DEFER, -x, key=lambda x: -x[0]), (x, f))
    else:
        DEFERF = f
        setitimer(ITIMER_REAL, x)


def deferok(*x):
    global DEFERF
    DEFERF()
    while DEFER and DEFER[-1][0] < time():
        DEFER.pop()[1]()
    if DEFER:
        x, DEFERF = DEFER.pop()
        setitimer(ITIMER_REAL, x - time())
    else:
        DEFERF = None


signal(SIGALRM, deferok)
for x in range(1, 10):
    print("start", x)
    defer(x, lambda x=x: print("end", x, flush=True))
```

```python
from random import choice
from signal import ITIMER_REAL, SIGALRM, SIGUSR1, pthread_kill, setitimer, signal
from threading import Thread, current_thread

from greenlet import greenlet


def get():
    return FIBERS.setdefault(current_thread(), {greenlet.getcurrent(): []})


def fiber(f):
    fiber = greenlet(f)
    get()[fiber] = []
    return fiber


def wait(*fibers):
    fiber = greenlet.getcurrent()
    get()[fiber].extend(fibers)
    while get()[fiber]:
        schedule()


def schedules(*_):
    print(end=f"schedule all {current_thread().name}\n")
    global FIBERS
    FIBERS = {i: j for i, j in FIBERS.items() if i.is_alive()}
    for i in FIBERS:
        if i != current_thread():
            pthread_kill(i.ident, SIGUSR1)
    schedule()


def schedule(*_):
    fibers = FIBERS[current_thread()] = {i: [j for j in j if not j.dead] for i, j in get().items() if not i.dead}
    fiber = choice([i for i, j in fibers.items() if not j])
    if fiber != greenlet.getcurrent():
        print(end=f"schedule {current_thread().name} {id(fiber):x}\n")
        fiber.switch()


def fmap(f, *a):
    threads = [Thread(target=fmap, args=(f, a)) for a in a[1:]]
    for i in threads:
        i.start()
    wait(*[fiber(lambda a=a: f(a)) for a in a[0]])
    for i in threads:
        i.join()


FIBERS = {}
setitimer(ITIMER_REAL, 1, 1)
signal(SIGALRM, schedules)
signal(SIGUSR1, schedule)
```
