```python
from copy import deepcopy
from re import findall

import ray.tune
from nn import *
from scipy.optimize import bisect


def arg(**_):
    o = {}
    for i, _ in _.items():
        for i in i.split("_"):
            t = o
            i = findall("([A-Za-z][a-z]*)([0-9]*)", i)
            for j, k in i[:-1]:
                if k:
                    if j not in t:
                        t[j] = [{}]
                    t[j] += [deepcopy(t[j][-1]) for k in range(int(k) + 1 - len(t[j]))]
                    t = t[j][int(k)]
                else:
                    if j not in t:
                        t[j] = {}
                    t = t[j]
            t[i[-1][0]] = _
    return o


class param(dict):
    def __mul__(s, x):
        return param({_: (s * x).__ceil__() if isinstance(s, float) else s for _, s in s.items()})


def decay(x, k, d=-4, shift=1, logspace=False):
    x = x[(..., slice(-shift), *[slice(None)] * (-d - 1))]
    k = k * torch.arange(x.size(d) / 2, -x.size(d) / 2, -1, dtype=x.dtype, device=x.device)[(..., *[None] * (-d - 1))]
    if logspace:
        x = x.sign().mul_(x.abs_().add_(5e-41).log_().add_(k).logcumsumexp(d).sub_(a.logcumsumexp(k)).exp_())
    else:
        x = x.mul_(k.exp_()).cumsum_(d).div_(k.cumsum_(d))
    return functional.pad(x, [*[0, 0] * (-d - 1), shift, 0])


def R(i, h, k, v, o, s=1):
    return Seq(
        Dot(
            Seq(
                Seq(Linear(i, h * k), Unflatten(-1, [h, 1, k])),
            ),
            Seq(
                Tup(
                    Mul(
                        Seq(Linear(i, h * k), Unflatten(-1, [h, 1, k])),
                        Seq(Linear(i, h * v, False), Unflatten(-1, [h, v, 1])),
                    ),
                    Add(
                        Const(Parameter(init.uniform_(torch.empty(h, 1, k), -0.5, -0.2))),
                        Const(Parameter(init.uniform_(torch.empty(h, v, 1), -0.5, -0.2))),
                    ),
                ),
                F(decay, shift=s, unpak=True),
            ),
        ),
        Flatten(-3),
        Linear(h * v, o, False),
    )


def C(i, h, o=0, a=0, r=3):
    return Seq(
        SepConvLeft(i, h, o or a, r),
        not o and Seq(BatchNorm(a), SiLU(True)),
    )


def B(i, h, o, **_):
    return Add(
        Seq(
            Add(*[Seq(*[eval(f)(**_) for _ in _]) for f, _ in _.items()]),
            BatchNorm(h),
            Linear(h, o, False),
        ),
        Linear(i, o, False),
    )


def Model(e, i, **_):
    return Seq(
        Embedding(e, i),
        *[eval(f)(**_) for f, _ in _.items() for _ in _],
    )


def train(_):
    dat = Dat("py.txt", 32, 256)
    _ = param(e_B1o=len(dat.unmap), i_B0i_B0R0i_B0C0i=1.0, **_)
    m = Model(**_ * bisect(lambda x: sum(_.numel() for _ in Model(**_ * x).parameters()) - 50000, 2, 128, xtol=0.1)).to("cuda", torch.bfloat16)
    o = torch.optim.Adam(m.parameters())
    O = torch.optim.lr_scheduler.OneCycleLR(o, 0.005, 10000, pct_start=0.01, div_factor=4, final_div_factor=9)
    X = dat()
    L = A = 0
    for _ in range(O.last_epoch + 1, O.total_steps + 1):
        x, X = X, dat()
        y = m(x[:, :-1])
        x = x[:, 1:]
        l = functional.cross_entropy(y.mT, x)
        l.backward()
        o.step()
        o.zero_grad()
        O.step()
        L += l.detach().float()
        A += (y.argmax(2) == x).sum()
        if not _ % 100:
            ray.train.report({"l": float(L), "a": float(A) / x.numel()})
            L = A = 0
```
